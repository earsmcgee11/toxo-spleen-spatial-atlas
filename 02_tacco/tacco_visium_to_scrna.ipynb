# TACCO (Figure 2D)

To relate the spatial regions to the single-cell data, we transferred spatial gene expression signatures from the Visium data onto individual cells using TACCO (Transfer of Annotations to Cells and Cell Objects) (Mages et al., 2023). This assigned each cell a similarity score for every spatial region, enabling us to infer its most likely anatomical localization (Fig. 2D). As expected, the spatial signature from the TZ and BZ clusters were highest in T cells and B cells respectively, while the 2o Fol signature was enriched in germinal center B cells.

Manuscript parameters (`tc.tl.annotate`): `normalize_to='reference'`; `bisections=20`, `bisection_divisor=19`; `assume_valid_counts=True`, `lamb=0`; `min_log2foldchange=0`, `min_genes_per_cell=1`.
